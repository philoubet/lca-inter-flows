```python
from geopolrisk.assessment.utils import (
    default_rmlist,
)  # This is a functions that returns the default list of metals that are available in the library database


ListofMetals = default_rmlist()
# fmt: off
ListofCountries = [36,40,56,76,124,152,156,180,251,276,380,508,528,554,579,697,699,710,716,757,826,842,
"NAFTA", "NPCC", "RAF", "RAS", "RER", "RLA", "RME", "RNA", "SAS", "UCTE", "WECC", "WEU"]

# List of countries to be assessed, you can use the ISO 3166-1 numeric code & "EU" for European Union

ListofYear = [
    2022
]  # List of years to be assessed, you can only use the year as an integer
```

```python
# fmt: on
from geopolrisk.assessment.utils import cvtcountry

Excl = []
for i in ListofCountries:
    try:
        cvtcountry(i)
    except Exception as e:
        print(f"Error converting countries: {e} + Country is {i}")
        Excl.append(i)
filteredListofCountries = [x for x in ListofCountries if x not in Excl]
```

## Transforming output to the format for the code


```python
import pandas as pd


# Load the data from the Excel file


df = pd.read_excel("CFs_all countries.xlsx")


CountryCodes = pd.read_csv("country_codes_V202501.csv")


mappingdf = pd.read_excel("final_geopolrisk_openlca_colored_verified.xlsx")


df_merged = df.merge(
    CountryCodes[["country_name", "country_iso2"]],
    left_on="Country [Economic Entity]",
    right_on="country_name",
    how="left",
)


df_merged["Country [Economic Entity]"] = df_merged["country_iso2"].combine_first(
    df_merged["Country [Economic Entity]"]
)


df_merged = df_merged.drop(columns=["country_name", "country_iso2"])


df_merged = df_merged.merge(
    mappingdf[
        ["ecoinvent Name", "Reference ID", "ecoinvent UUID", "Color", "cas", "category"]
    ],
    left_on="Raw Material",
    right_on="Reference ID",
    how="left",
)


df_merged = df_merged.rename(
    columns={
        "ecoinvent UUID": "ecoinvent_UUID",
        "cas": "cas_number",
        "category": "flow_category",
    }
)


# df_merged = df_merged.drop(columns=["Reference ID"])


df_merged["comp1"] = "Raw"


df_merged["comp2"] = "in ground"


df_merged["Unit"] = "kg Cu eq/kg"


export_df = df_merged[
    [
        "comp1",
        "comp2",
        "ecoinvent Name",
        "Reference ID",
        "Color",
        "GeoPolRisk Characterization Factor [eq. Kg-Cu/Kg]",
        "Country [Economic Entity]",
        "ecoinvent_UUID",
        "cas_number",
        "Unit",
    ]
].rename(
    columns={
        "ecoinvent Name": "Substance",
        "Reference ID": "Matched Substance",
        "GeoPolRisk Characterization Factor [eq. Kg-Cu/Kg]": "Supplyrisk",
        "Country [Economic Entity]": "Geography",
    }
)


export_df.to_csv(
    "model_csv_geopolrisk_with_colors_corrected_AK2.csv", sep=";", index=False
)
```
